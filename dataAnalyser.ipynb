```python
import pandas as pd 
import matplotlib.pyplot as plt
df = pd.read_csv('Dataset/go_emotions_dataset.csv')
df.head()
emotions= df.columns[3:]
print(emotions)


data = emotions.value_counts()
values = []

for emotion in emotions:
    # print(df[emotion].sum())
    values.append(emotion)

plt.pie(values , labels = emotions)
plt.axis('equal')
plt.title("Emotion Chart in the dataset")
plt.show()
```

```python
import numpy as np
# Analyse more 
value_array = np.array(values)
# Percent of emotion in the dataset with maximum count , output : 80% of dataset is of neutral type 
print(np.max(value_array) / np.sum(value_array))
```

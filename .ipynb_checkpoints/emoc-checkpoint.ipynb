```python
import numpy as np
import pandas as pd
import nltk
from sklearn.preprocessing import OneHotEncoder

data = pd.read_csv('/content/drive/MyDrive/Viora/emotion_sentimen_dataset_small.csv')
# data.head()
df = data["text"].to_frame(name= "text")
df.head()


# def normalizeData(df):
#     for index,row in df.iterrows():
#         text = row['text']
#         # text = text.lower()
#         # text.strip()
#         # text.strip('''
#         # !"#$%&'()*+,-./:;<=>?@[]\\^_`{|}~
#         # ''')
#         row['text'] = text
#     return df

# normalizeData(df)


def wordTokenize(text:str):
    return nltk.word_tokenize(text)

# df = normalizeData(df)

vocabulary = {
    "<PAD>": 0,
    "<UNK>": 1}

currIdx=2

max_len = 0

def buildVocabulary(df):
    global currIdx
    global max_len
    for idx,row in df.iterrows():
        text = row['text']
        text = wordTokenize(text)
        numSequence = []

        for word in text:
            if word not in vocabulary:
                vocabulary[word] = currIdx
                numSequence.append(currIdx)
                currIdx+=1
            else:
                numSequence.append(vocabulary[word])
        row['text'] = np.array(numSequence)
        max_len = max(max_len , len(numSequence))

buildVocabulary(df)
df.head()
```
